```python
day = "Saturday"

match day.lower():
    case "monday":
        print("Start of the work week!")
    case "wednesday":
        print("Midweek already!")
    case "friday":
        print("Almost weekend!")
    case "saturday" | "sunday":
        print("Weekend time!")
    case _:
        print("Just another day")
```
